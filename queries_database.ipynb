## weekday, weekend, holiday

```python
con.execute("""
CREATE VIEW tlc AS SELECT * FROM 'data/processed/tlc/yellow_zone_date_2025.parquet'
""")

con.execute("""
CREATE VIEW poi AS SELECT * FROM 'data/processed/poi/poi_zone_features.parquet'
""")

con.execute("""
CREATE VIEW weather AS SELECT * FROM 'data/processed/weather/weather_2025.csv'
""")
```

```python
con.execute("""
CREATE OR REPLACE TABLE holidays_2025 AS
SELECT * FROM (VALUES
    (DATE '2025-01-01', 'New Year''s Day'),
    (DATE '2025-01-20', 'Martin Luther King Jr. Day'),
    (DATE '2025-02-17', 'Presidents'' Day'),
    (DATE '2025-05-26', 'Memorial Day'),
    (DATE '2025-06-19', 'Juneteenth'),
    (DATE '2025-07-04', 'Independence Day'),
    (DATE '2025-09-01', 'Labor Day'),
    (DATE '2025-10-13', 'Columbus Day'),
    (DATE '2025-11-11', 'Veterans Day'),
    (DATE '2025-11-27', 'Thanksgiving Day'),
    (DATE '2025-12-24', 'Christmas Eve'),
    (DATE '2025-12-25', 'Christmas Day')
) AS t(date, holiday_name);
""")
```

```python
con.execute("""
CREATE OR REPLACE VIEW tlc_with_day_tag AS
SELECT
    t.*,
    CASE
        WHEN h.date IS NOT NULL THEN 'holiday'
        WHEN EXTRACT('dow' FROM CAST(t.date AS DATE)) IN (0, 6) THEN 'weekend'
        ELSE 'weekday'
    END AS day_tag,
    h.holiday_name
FROM tlc t
LEFT JOIN holidays_2025 h
ON CAST(t.date AS DATE) = h.date;
""")
```

```python
con.execute("""
CREATE OR REPLACE VIEW zone_tag_pickup_counts AS
SELECT
    taxi_zone_id,
    day_tag,
    SUM(trips_pickup) AS total_pickups,
    AVG(trips_pickup) AS mean_daily_pickups,
    COUNT(DISTINCT CAST(date AS DATE)) AS num_days
FROM tlc_with_day_tag
GROUP BY taxi_zone_id, day_tag;
""")
```

```python
con.execute("""
SELECT *
FROM zone_tag_pickup_counts
ORDER BY taxi_zone_id, day_tag
LIMIT 20;
""").df()
```

```python
con.execute("""
CREATE OR REPLACE VIEW zone_tag_pickup_wide AS
SELECT
    taxi_zone_id,
    MAX(CASE WHEN day_tag = 'weekday' THEN mean_daily_pickups END) AS mean_pickups_weekday,
    MAX(CASE WHEN day_tag = 'weekend' THEN mean_daily_pickups END) AS mean_pickups_weekend,
    MAX(CASE WHEN day_tag = 'holiday' THEN mean_daily_pickups END) AS mean_pickups_holiday
FROM zone_tag_pickup_counts
GROUP BY taxi_zone_id;
""")
```

```python
zone_stats = con.execute("""
SELECT *
FROM zone_tag_pickup_wide
""").df()
```

```python
## alternative
con.execute("""
CREATE OR REPLACE VIEW zone_weekend_holiday_lift AS
SELECT
    taxi_zone_id,
    mean_pickups_weekday,
    mean_pickups_weekend,
    mean_pickups_holiday,
    mean_pickups_weekend / NULLIF(mean_pickups_weekday, 0) AS weekend_lift,
    mean_pickups_holiday / NULLIF(mean_pickups_weekday, 0) AS holiday_lift
FROM zone_tag_pickup_wide;
""")
```

```python
# visualizaiton 1
def plot_pickup_map_by_tag(zones, zone_tag, tag, output_path=None):
    """
    Choropleth map of mean daily pickup count by taxi zone for one day tag.
    tag: 'weekday', 'weekend', or 'holiday'
    """

    df_tag = zone_tag[zone_tag["day_tag"] == tag].copy()

    gdf = zones.merge(
        df_tag,
        on="taxi_zone_id",
        how="left"
    )

    fig, ax = plt.subplots(figsize=(10, 8))

    gdf.plot(
        column="mean_daily_pickups",
        ax=ax,
        legend=True,
        missing_kwds={
            "color": "lightgrey",
            "label": "No data"
        }
    )

    ax.set_title(
        f"Mean Daily Taxi Pickups by Zone: {tag.capitalize()}",
        fontsize=14
    )
    ax.axis("off")

    if output_path:
        plt.savefig(output_path, dpi=300, bbox_inches="tight")

    plt.show()
```

```python
plot_pickup_map_by_tag(
    zones,
    zone_tag,
    "weekday",
    "outputs/map_pickups_weekday.png"
)

plot_pickup_map_by_tag(
    zones,
    zone_tag,
    "weekend",
    "outputs/map_pickups_weekend.png"
)

plot_pickup_map_by_tag(
    zones,
    zone_tag,
    "holiday",
    "outputs/map_pickups_holiday.png"
)
```

### visualization 2

```python
def plot_pickup_boxplot_by_tag(zone_tag, output_path=None):
    """
    Box plot of zone-level mean daily pickups by day tag.
    Each observation is one taxi zone under one tag.
    """

    plot_df = zone_tag.copy()

    tag_order = ["weekday", "weekend", "holiday"]
    data = [
        plot_df.loc[plot_df["day_tag"] == tag, "mean_daily_pickups"].dropna()
        for tag in tag_order
    ]

    fig, ax = plt.subplots(figsize=(8, 6))

    ax.boxplot(
        data,
        labels=[tag.capitalize() for tag in tag_order],
        showfliers=True
    )

    ax.set_title("Distribution of Mean Daily Taxi Pickups by Day Type")
    ax.set_xlabel("Day Type")
    ax.set_ylabel("Mean Daily Pickups per Zone")

    if output_path:
        plt.savefig(output_path, dpi=300, bbox_inches="tight")

    plt.show()

plot_pickup_boxplot_by_tag(
    zone_tag,
    "outputs/boxplot_pickups_by_day_tag.png"
)
```

### dealing with skewed spatial count

introduce fixed color scale and log transformation

```python
zone_tag["log_mean_daily_pickups"] = np.log1p(zone_tag["mean_daily_pickups"])

global_vmin = zone_tag["log_mean_daily_pickups"].min()
global_vmax = zone_tag["log_mean_daily_pickups"].max()

def plot_pickup_map_by_tag_log(zones, zone_tag, tag, output_path=None):
    df_tag = zone_tag[zone_tag["day_tag"] == tag].copy()

    gdf = zones.merge(
        df_tag,
        on="taxi_zone_id",
        how="left"
    )

    fig, ax = plt.subplots(figsize=(10, 8))

    gdf.plot(
        column="log_mean_daily_pickups",
        ax=ax,
        legend=True,
        vmin=global_vmin,
        vmax=global_vmax,
        missing_kwds={
            "color": "lightgrey",
            "label": "No data"
        },
        legend_kwds={
            "label": "log(1 + mean daily pickups)"
        }
    )

    ax.set_title(
        f"Mean Daily Taxi Pickups by Zone: {tag.capitalize()}",
        fontsize=14
    )
    ax.axis("off")

    if output_path:
        plt.savefig(output_path, dpi=300, bbox_inches="tight")

    plt.show()

for tag in ["weekday", "weekend", "holiday"]:
    plot_pickup_map_by_tag_log(
        zones,
        zone_tag,
        tag,
        f"outputs/map_pickups_{tag}_log_fixed_scale.png"
    )
```

### interactive map

```python
con.execute("""
CREATE OR REPLACE VIEW zone_tag_pickup_stats AS
SELECT
    taxi_zone_id,
    day_tag,
    SUM(trips_pickup) AS total_pickups,
    AVG(trips_pickup) AS mean_daily_pickups,
    MIN(trips_pickup) AS min_daily_pickups,
    MAX(trips_pickup) AS max_daily_pickups,
    COUNT(DISTINCT CAST(date AS DATE)) AS num_days
FROM tlc_with_day_tag
GROUP BY taxi_zone_id, day_tag;
""")

zone_tag_stats = con.execute("""
SELECT *
FROM zone_tag_pickup_stats
""").df()
```

```python
import folium

def make_interactive_pickup_map(zones, zone_tag_stats, tag, output_html=None):
    df_tag = zone_tag_stats[zone_tag_stats["day_tag"] == tag].copy()
    df_tag["log_mean_daily_pickups"] = np.log1p(df_tag["mean_daily_pickups"])

    gdf = zones.merge(df_tag, on="taxi_zone_id", how="left")
    gdf = gdf.to_crs("EPSG:4326")

    m = folium.Map(
        location=[40.7128, -74.0060],
        zoom_start=10,
        tiles="cartodbpositron"
    )

    folium.Choropleth(
        geo_data=gdf,
        data=gdf,
        columns=["taxi_zone_id", "log_mean_daily_pickups"],
        key_on="feature.properties.taxi_zone_id",
        fill_opacity=0.75,
        line_opacity=0.2,
        legend_name="log(1 + mean daily pickups)"
    ).add_to(m)

    tooltip_fields = [
        "taxi_zone_id",
        "zone",
        "borough",
        "day_tag",
        "mean_daily_pickups",
        "min_daily_pickups",
        "max_daily_pickups",
        "total_pickups",
        "num_days"
    ]

    tooltip_aliases = [
        "Taxi Zone ID:",
        "Zone:",
        "Borough:",
        "Day Tag:",
        "Mean Daily Pickups:",
        "Min Daily Pickups:",
        "Max Daily Pickups:",
        "Total Pickups:",
        "Number of Days:"
    ]

    folium.GeoJson(
        gdf,
        name=f"{tag} pickup stats",
        tooltip=folium.GeoJsonTooltip(
            fields=[f for f in tooltip_fields if f in gdf.columns],
            aliases=tooltip_aliases[:len([f for f in tooltip_fields if f in gdf.columns])],
            localize=True,
            sticky=True
        ),
        style_function=lambda feature: {
            "fillOpacity": 0,
            "color": "black",
            "weight": 0.2
        }
    ).add_to(m)

    folium.LayerControl().add_to(m)

    if output_html:
        m.save(output_html)

    return m

m_weekday = make_interactive_pickup_map(
    zones,
    zone_tag_stats,
    "weekday",
    "outputs/interactive_pickups_weekday.html"
)

m_weekday
```

```python
for tag in ["weekday", "weekend", "holiday"]:
    make_interactive_pickup_map(
        zones,
        zone_tag_stats,
        tag,
        f"outputs/interactive_pickups_{tag}.html"
    )
```
